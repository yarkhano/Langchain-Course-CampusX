## Models(Language & embedding models)-> It is an interface for connecting to models in langchain. ##

## Language Models ##

**We have LLM and chat models ,langchain is decreasing its support for LLMs while encourage to use chat models in your apps. LLm->General purpose models for text generation,code ,summarization and translation. Chat Models->Fine tuned for conversation and are good than LLMs.use in chatbots,assistants,ai tutors.**

```python
#openai llm(not recommended)

from langchain_openai import OpenAI
from dotenv import load_dotenv
load_dotenv()


llm = OpenAI(model='gpt-3.5-turbo-instruct')
result = llm.invoke("The name of the capital of pakistan is?")
print(result.content)
```

## Chat Models ##

**temperature define creativity or determinism,how much you want change or creativeness from your actual answer.**

```python
#openAI

from langchain_openai import ChatOpenAI
from dotenv import load_dotenv
load_dotenv()


model = ChatOpenAI(model='gpt-4.5',temperature=0,max_completion_tokens=10)
result = model.invoke("The name of the capital of pakistan is?")
print(result.content)
```

```python
#claude

from langchain_anthropic import ChatAnthropic
from dotenv import load_dotenv
load_dotenv()


model = ChatAnthropic(model='claude:...')
result = model.invoke("The name of the capital of pakistan is?")
print(result.content)
```

```python
#google chat model
from langchain_google_genai import ChatGoogleGenerativeAI

from dotenv import load_dotenv
load_dotenv()

model = ChatGoogleGenerativeAI(model="gemini-2.5-flash")
result = model.invoke("The name of the capital of Pakistan is?")
print(result.content)
```

## Open Source Chat Models ##

**Hugging Face is the biggest repository for open source Ai models**

```python
#huggingface
from langchain_huggingface import ChatHuggingFace,HuggingFaceEndpoint
import os
from dotenv import load_dotenv
load_dotenv()

hug_token = os.getenv("HUGGINGFACEHUB_ACCESS_TOKEN")
llm = HuggingFaceEndpoint(repo_id="Qwen/Qwen2.5-7B-Instruct",task="text-generation",huggingfacehub_api_token=hug_token,max_new_tokens=100)

model = ChatHuggingFace(llm=llm)

result = model.invoke("The name of the capital of Pakistan is?")
print(result.content)
```

** downloading model to local machine **

```python
from langchain_huggingface import ChatHuggingFace , HuggingFacePipeline

llm_hug = HuggingFacePipeline.from_model_id(model_id="Qwen/Qwen2.5-0.5B-Instruct",
        task="text-generation",
        pipeline_kwargs=dict(
        temperature=0.3,
        max_new_tokens=100,
        ))

model = ChatHuggingFace(llm=llm)

result = model.invoke("The name of the capital of Pakistan is?")
print(result.content)
```
