**built in tools**

```python
from langchain_community.tools import DuckDuckGoSearchRun

search_tool = DuckDuckGoSearchRun()

results = search_tool.invoke("current weather of Islamabad")
print(results)
```

**Custom Tools**

**@tool make the function to communicate with LLM.**

```python
from langchain_core.tools import tool

def multiply(a,b):
    """This function multiply two numbers"""
    return a*b

def multiply(a:int,b:int)->int:
    """This function multiply two numbers"""
    return a*b

@tool
def multiply(a:int,b:int)->int:
    """This function multiply two numbers"""
    return a*b

result = multiply.invoke({"a":3,"b":4})
print(result)
```

## Structured tool->Where we use the pydantic for tool creation. ##

## BaseTool->Abstract class for all tools in langchain.All other tools like @tool and structured tools are built on the top of the basetool. ##

```python
from langchain.tools import BaseTool
from typing import Type
from pydantic import  BaseModel , Field


class MultimplyInput(BaseModel):
    a: int = Field(required=True, description="this is a attribute")
    b: int = Field(required=True, description="this is b second attribute")

class MultiplyNums(BaseTool):
    name: str = "multiply"
    description :str = "multiply two numbers"

    args_schema: Type[BaseModel] = MultimplyInput

    def _run(self,a: int, b: int) -> int:
        return a*b
```

```python
m_tool = MultiplyNums()
result = m_tool.invoke({"a":3,"b":4})
print(result)
```
