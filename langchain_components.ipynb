**langchain is a model agnostic framework->if you later want to change the llm model you have to change a single line of code or very less number of lines**

**In chains output of one thing become input of other**

**Minimal code is required such a complex tasks**

## Components Of Langchain ##

**1)Models-> Component use for interacting with LLMs.
We do no need to change complete code for every llm just two lines by using langchain models interface.We have language models and embedding models(help in semantic search)**

**LLM has two beneficial things->NLU+Context aware text generation.**

**2)Prompt->Input given to LLM.We can implement genral prompts,role based as well as few shot prompts(giving input output samples)**

**3)Chains->Use to build pipeline.Where he output previous step become automatically input of next step.Use to connect componenets**

**4)Indexes->Consists of Document loader,textsplitter,adding to vector store and retriever.So indexes is where you can built llm base apps which are connected to external knowledge source(websites,pdfs,databases).**

**Memmory-> LLM API calls are stateless(the current api call or prompt do not know about previous one).We ave 4 types of memmory in langchain**

**5)Agents->have resoning capability(breakdown and do things step by step) and connected tools.**
