## LLMs are good at reasing and speaking but can not do work on ist own ,tools are function which has capability to connect with LLM and perform action. ##

**Tool Binding->Which tools available description and other details.**

```python
from langchain_core.tools import tool
from langchain_core.messages import HumanMessage
import requests
import os
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
GOOGLE_API_KEY = os.getenv("GOOGLE_API_KEY")
model = ChatGoogleGenerativeAI(model="gemini-2.5-flash",api_key=GOOGLE_API_KEY)
```

```python
@tool
def multiply(a:int ,b:int)->int:
    """multiply two numbers"""
    return a * b
```

```python
m_tool = model.bind_tools([multiply])
```
