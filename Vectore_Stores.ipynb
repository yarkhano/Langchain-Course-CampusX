## Vector Stores->Store the semantic meaning of the text in numbers ##

**Angular Distance use to get similarity between vectors**

**In RDBMS we can not do similarity search,and have no feature for semantic search.**

**Vectore store provide-> Storage(disk or in memmory),similarity search,indexing,CRUD operations.**

**Indexing->Using technique or algorithms to make fast the similarity search. e.g a query is arrived to vector store and in vector store we ave 10 lakh vectors so it is very time consuming to compare query with 10lakh vectors so we create clusters and and then average of every cluster is being find and with average of each cluster the query is compared for similarity search.**

**Vectore Store Vs Vector Database->Vector Store has features of Storage and search. While vector database has features of vector store as well as the additional database like features such as ditributed architecture,backup & restore,role based access etc.**

```python
from langchain.schema import Document
from langchain_chroma import Chroma
from langchain_openai import OpenAIEmbeddings

# 1. Create LangChain documents for IPL players
doc1 = Document(
    page_content="Virat Kohli is one of the most successful and consistent batsmen in IPL history. Known for his aggressive batting style and leadership.",
    metadata={"team": "Royal Challengers Bangalore"}
)

doc2 = Document(
    page_content="Rohit Sharma is the most successful captain in IPL history, leading Mumbai Indians to five titles. He's known for his calm and composed batting.",
    metadata={"team": "Mumbai Indians"}
)

doc3 = Document(
    page_content="MS Dhoni, famously known as Captain Cool, has led Chennai Super Kings to multiple IPL titles. His finishing skills, wicket-keeping, and sharp mind make him legendary.",
    metadata={"team": "Chennai Super Kings"}
)

doc4 = Document(
    page_content="Jasprit Bumrah is considered one of the best fast bowlers in T20 cricket. Playing for Mumbai Indians, he is known for his pinpoint yorkers and economy.",
    metadata={"team": "Mumbai Indians"}
)

doc5 = Document(
    page_content="Ravindra Jadeja is a dynamic all-rounder who contributes with both bat and ball. Representing Chennai Super Kings, his quick fielding and left-arm spin are invaluable.",
    metadata={"team": "Chennai Super Kings"}
)

docs = [doc1, doc2, doc3, doc4, doc5]

# 2. Initialize Chroma vector store
vector_store = Chroma(
    embedding_function=OpenAIEmbeddings(),
    persist_directory='my_chroma_db',  # or 'chroma_db'
    collection_name='sample'
)

# 3. Add documents to vector store
vector_store.add_documents(docs)

# 4. Search documents (Similarity Search)
results = vector_store.similarity_search(
    query='Who among these are a bowler?',
    k=2
)
print("Similarity Search Results:", results)

# 5. Metadata filtering
filtered_results = vector_store.similarity_search_with_score(
    query="",
    filter={"team": "Chennai Super Kings"}
)
print("Filtered Results:", filtered_results)

# 6. Update documents
updated_doc1 = Document(
    page_content="Virat Kohli, the former captain of Royal Challengers Bangalore (RCB), is renowned for his aggressive leadership and consistency.",
    metadata={"team": "Royal Challengers Bangalore"}
)

# Note: Replace the ID below with the actual document ID returned when adding the document
vector_store.update_document(
    document_id='fdb1dc2c-b27b-4f18-80ef-ee48f4d9a31f',
    document=updated_doc1
)
```
